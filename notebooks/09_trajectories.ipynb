# Trajectories

This notebook computes trajectory existence and probability for the SD Markov chain across Boolean network models.

Results are cached to CSV to allow rapid iteration on figures.

Generates figure s8.

```python
import pandas as pd

import sdmarkov.plotting.config as pc
from sdmarkov.graph import get_markov_chain
from sdmarkov.paths import get_all_paths, get_markov_chain_path_probs, get_stg_path_probs, get_random_path_probs
from sdmarkov.paths import compare_path_reachability, compare_path_rmsd
from sdmarkov.analysis.model_context import iter_model_contexts
from sdmarkov.analysis.collect import generate_dataframe
from sdmarkov.analysis.postprocessing import add_classification_metrics, network_level_metrics_df
from sdmarkov.plotting.figures import plot_distribution_figure, save_figure
```

```python
IMPORT_CSV = False
REBUILD_CONTEXTS = False
DEBUG = False

CACHE_DIR = "../.cache/model_contexts"
CSV_FILE = "../results/trajectories.csv"

# Parameters for building contexts
MODEL_DIRECTORY = "../models/random_nk3"
UPDATE = "asynchronous"
N_RANDOM = 30 # number of random models
NSQUARE_STEPS = 20 # number of steps for nsquare

# Figure configuration
FIGURE_DIR = "../results/figures"

SUP_FIGURE_ID = "s8"
METRICS = ["rmsd"]
METHODS = None # plot all methods
```

```python
def get_data_trajectories(
    context,
    DEBUG=False,
):
    """
    Compare trajectory/path-level properties between the STG
    and grouped Markov-chain representations.

    Returns
    -------
    pd.DataFrame
        Long-format table with columns:
        bnet, method, run, paths_reachability, paths_rmsd
    """

    rows = []

    # Unpack model context
    model_name = context.model_name
    update = context.update
    n_states = context.n_states
    n_attractors = context.n_attractors
    stg = context.stg
    T = context.T
    sd_indices = context.sd_indices
    Tsd = context.Tsd

    # filter out fully percolated models
    if n_states == 1:
        return pd.DataFrame(rows)

    sd_mc = get_markov_chain(Tsd, sd_indices, DEBUG=DEBUG)
    sd_all_paths = get_all_paths(sd_mc, DEBUG=DEBUG)

    sd_mc_path_probs = get_markov_chain_path_probs(sd_mc, sd_all_paths, DEBUG=DEBUG)
    sd_stg_path_probs = get_stg_path_probs(sd_all_paths, sd_indices, stg, T, DEBUG=DEBUG)

    # --- helper: add row ---
    def _add_row(method, stg_probs, predictions, run=None):
        TP, FP, TN, FN = compare_path_reachability(stg_probs, predictions, type="non_edges", DEBUG=DEBUG)
        rmsd_val = compare_path_rmsd(stg_probs, predictions, type="non_edges", DEBUG=DEBUG)
        rows.append({
            "model": model_name,
            "method": method,
            "run": run,
            "rmsd": rmsd_val,
            "TP": TP,
            "FP": FP,
            "TN": TN,
            "FN": FN,
            "n_states": n_states,
            "n_attractors": n_attractors,
            "update_scheme": update
        })

    # --- deterministic SD method ---
    _add_row("sd_mc", sd_stg_path_probs, sd_mc_path_probs)

    # --- randomized baselines ---
    for i in range(N_RANDOM):
        random_path_probs = get_random_path_probs(sd_mc_path_probs, seed=i)
        _add_row("ref", sd_stg_path_probs, random_path_probs, run=i)

    return pd.DataFrame(rows)
```

```python
if IMPORT_CSV:
    df = pd.read_csv(CSV_FILE)
    print(f"Loaded cached results from {CSV_FILE}.")
else:
    ctxs = iter_model_contexts(model_directory=MODEL_DIRECTORY,
                               update=UPDATE,
                               nsquare_steps=NSQUARE_STEPS,
                               n_random=N_RANDOM,
                               cache_dir=CACHE_DIR,
                               force_rebuild=REBUILD_CONTEXTS,
                               DEBUG=DEBUG,
                               )

    df = generate_dataframe(
        model_contexts=ctxs,
        data_function=get_data_trajectories,
        schema={
            "TP": int,
            "FP": int,
            "TN": int,
            "FN": int,
            "n_states": int,
            "n_attractors": int,
            "run": "Int64",
        },
        DEBUG=DEBUG,
    )
    df.to_csv(CSV_FILE, index=False)
    print(f"Saved results to {CSV_FILE}.")
```

```python
print("Preview of results DataFrame:")
display(df.head())

print("\nDataFrame shape:", df.shape)
print("Methods present:", df.method.unique())
print("Number of unique networks:", df.model.nunique())
```

## Data analysis

```python
# filter models with TP+FP > 0
df = df[(df.TP + df.FP) > 0]

# add precision, recall, specificity, and NPV
df = add_classification_metrics(df)

print("Added classification metrics")
print("DataFrame shape:", df.shape)
print("Methods present:", df.method.unique())
print("Number of unique networks:", df.model.nunique())

display(df.head())
```

```python
df_metrics = network_level_metrics_df(df=df)

print("Preview of network-level metrics DataFrame:")
display(df_metrics.head())

print("\nDataFrame shape:", df_metrics.shape)
print("Methods present:", df_metrics.method.unique())
print("Number of unique networks:", df_metrics.model.nunique())

print("\nNetwork-level classification metrics:")
metrics = ["rmsd", "precision", "recall", "specificity", "npv"]
summary = (
    df_metrics
    .groupby("method")[metrics]
    .agg(["mean", "min", "median", "max"])
)

display(summary)
```
