# Decision transitions

This notebook identifies decision transitions and compare metrics for SD, null, and random groupings across Boolean network models.

Results are cached to CSV to allow rapid iteration on figures.

Generates figure s7.

```python
import pandas as pd

import sdmarkov.plotting.config as pc
from sdmarkov.decisions import get_decision_matrix, expand_decision_matrix, compare_decision_matrices
from sdmarkov.analysis.model_context import iter_model_contexts
from sdmarkov.analysis.collect import generate_dataframe
from sdmarkov.analysis.postprocessing import add_classification_metrics, network_level_metrics_df
from sdmarkov.plotting.figures import plot_distribution_figure, save_figure
```

```python
IMPORT_CSV = False
REBUILD_CONTEXTS = False
DEBUG = False

CACHE_DIR = "../.cache/model_contexts"
CSV_FILE = "../results/decision_transitions.csv"

# Parameters for building contexts
MODEL_DIRECTORY = "../models/random_nk3"
UPDATE = "asynchronous"
N_RANDOM = 30 # number of random models
NSQUARE_STEPS = 20 # number of steps for nsquare

# Figure configuration
FIGURE_DIR = "../results/figures"

SUP_FIGURE_ID = "s7"
METRICS = ["precision", "recall", "specificity", "npv"]
METHODS = None # plot all methods
```

## Data generation or loading

```python
# --- Experiment-specific analysis ---
def get_data_decision_transitions(
    context,
    DEBUG=False,
):
    """
    Compare decision transition matrices between full, grouped,
    and random groupings of a Boolean network.

    Parameters
    ----------
    context : ModelContext
        Pre-built model context.
    DEBUG : bool, optional
        Enable verbose output in lower-level metric computations.

    Returns
    -------
    pd.DataFrame
        Long-format DataFrame with one row per method (and per random run).
        Returns an empty DataFrame if the model has fewer than two attractors.
    """

    rows = []

    # Unpack model context
    model_name = context.model_name
    update = context.update
    n_states = context.n_states
    n_attractors = context.n_attractors
    n_random = context.random.n_random
    T = context.T
    sd_indices = context.sd_indices
    Tsd = context.Tsd
    null_indices = context.null_indices
    Tnull = context.Tnull
    random_indices_list = context.random.indices
    Trandom_list = context.random.Trandom

    # filter out networks with less than 2 attractors
    if n_attractors < 2:
        return pd.DataFrame(rows)

    # Actual
    decisions = get_decision_matrix(T, DEBUG=DEBUG)

    # MC
    sd_decisions = get_decision_matrix(Tsd, DEBUG=DEBUG)
    sd_decisions_expanded = expand_decision_matrix(
        sd_decisions, sd_indices, DEBUG=DEBUG        
    )
    null_decisions = get_decision_matrix(Tnull, DEBUG=DEBUG)
    null_decisions_expanded = expand_decision_matrix(
        null_decisions, null_indices, DEBUG=DEBUG        
    )

    # --- helper: add row ---
    def _add_row(method, predictions, run=None):
        TP, FP, TN, FN = compare_decision_matrices(
            decisions, predictions, DEBUG=DEBUG
        )
        rows.append({
            "model": model_name,
            "method": method,
            "run": run,
            "TP": TP,
            "FP": FP,
            "TN": TN,
            "FN": FN,
            "n_states": n_states,
            "n_attractors": n_attractors,
            "update_scheme": update
        })

    # --- deterministic methods ---
    _add_row("sd_mc", sd_decisions_expanded)
    _add_row("null_mc", null_decisions_expanded)

    # --- random baseline ---
    for i in range(n_random):
        random_indices = random_indices_list[i]
        Trandom = Trandom_list[i]
        random_decisions = get_decision_matrix(Trandom, DEBUG=DEBUG)
        random_decisions_expanded = expand_decision_matrix(
            random_decisions, random_indices, DEBUG=DEBUG
        )
        _add_row("random_mc", random_decisions_expanded, run=i)

    return pd.DataFrame(rows)
```

```python
if IMPORT_CSV:
    df = pd.read_csv(CSV_FILE)
    print(f"Loaded cached results from {CSV_FILE}.")
else:
    ctxs = iter_model_contexts(model_directory=MODEL_DIRECTORY,
                               update=UPDATE,
                               nsquare_steps=NSQUARE_STEPS,
                               n_random=N_RANDOM,
                               cache_dir=CACHE_DIR,
                               force_rebuild=REBUILD_CONTEXTS,
                               DEBUG=DEBUG,
                               )

    df = generate_dataframe(
        model_contexts=ctxs,
        data_function=get_data_decision_transitions,
        schema={
            "TP": int,
            "FP": int,
            "TN": int,
            "FN": int,
            "n_states": int,
            "n_attractors": int,
            "run": "Int64",
        },
        DEBUG=DEBUG,
    )
    df.to_csv(CSV_FILE, index=False)
    print(f"Saved results to {CSV_FILE}.")
```

```python
print("Preview of results DataFrame:")
display(df.head())

print("\nDataFrame shape:", df.shape)
print("Methods present:", df.method.unique())
print("Number of unique networks:", df.model.nunique())
```

## Data analysis

```python
# filter out networks with a single attractor
df_multi = df[df["n_attractors"] > 1].copy()

# add precision, recall, specificity, and NPV
df_multi = add_classification_metrics(df_multi)

print("\nAdded classification metrics")
print("DataFrame shape:", df_multi.shape)
print("Methods present:", df_multi.method.unique())
print("Number of unique networks:", df_multi.model.nunique())

display(df_multi.head())
```

```python
df_metrics = network_level_metrics_df(df=df_multi)

print("Preview of network-level metrics DataFrame:")
display(df_metrics.head())

print("\nDataFrame shape:", df_metrics.shape)
print("Methods present:", df_metrics.method.unique())
print("Number of unique networks:", df_metrics.model.nunique())

print("\nNetwork-level classification metrics:")
metrics = ["precision", "recall", "specificity", "npv"]
summary = (
    df_metrics
    .groupby("method")[metrics]
    .agg(["mean", "min", "median", "max"])
)

display(summary)
```
